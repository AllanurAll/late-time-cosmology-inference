# 08 — CC + BAO: flat XCDM

Flat XCDM extends flat $\Lambda$CDM by replacing the cosmological constant with a homogeneous dark-energy component having constant equation-of-state parameter $w_X$. The background is

$$E^2(z)=\Omega_m(1+z)^3+(1-\Omega_m)(1+z)^{3(1+w_X)}.$$

The $\Lambda$CDM limit is exactly $w_X=-1$. The sampled parameters are $h$, $\Omega_bh^2$, $\Omega_ch^2$, and $w_X$. The primary reported triangle uses $H_0$, $\Omega_m$, and $w_X$.
