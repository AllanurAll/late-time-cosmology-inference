Pantheon+ is used here as an uncalibrated relative-distance probe. The additive supernova intercept (the degenerate combination of absolute magnitude and $H_0$) is marginalized analytically. Therefore the Pantheon+-only flat-$\Lambda$CDM analysis constrains $\Omega_m$, not an independent $H_0$.

Two selections are retained: the public Pantheon+ likelihood selection $z_{HD}>0.01$ and a stricter cross-check that additionally excludes SH0ES calibrator-host rows.
