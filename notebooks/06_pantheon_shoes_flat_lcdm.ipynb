# 06 — Pantheon+ + SH0ES calibrated flat $\Lambda$CDM

This stage follows the public Pantheon+SH0ES likelihood. The selection is $(z_{HD}>0.01)\ \mathrm{OR}\ \mathrm{IS\_CALIBRATOR}$; Cepheid-host supernovae are compared to `CEPH_DIST`, while the remaining supernovae are compared to the flat-$\Lambda$CDM luminosity distance. The global SN absolute magnitude $M_B$ is marginalized analytically.

Unlike uncalibrated Pantheon+ alone, the Cepheid host distances break the $M_B$--$H_0$ degeneracy, so $H_0$ is now an identifiable cosmological parameter.
