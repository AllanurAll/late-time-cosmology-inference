1. a fixed $r_s=147.4$ Mpc reproduction of the older analysis;
2. a physical-density parameterization using the Aubourg sound-horizon approximation and the broad priors of Cao, Ryan & Ratra (2021).

The second BAO-only $H_0$ posterior is explicitly prior-conditioned because a long physical-density degeneracy remains.
