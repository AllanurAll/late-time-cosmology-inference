# 00 — Data and covariance audit

This notebook checks the CC covariance construction and the BAO likelihood before any MCMC sampling.


```python
from pathlib import Path
import numpy as np
from late_time_cosmology.cc import load_cc32, build_cc_covariance

root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df, sys = load_cc32(root/'data'/'cc')
cov = build_cc_covariance(df, sys)
print('N_CC =', len(df))
print('N_correlated =', int(df['moresco_correlated'].sum()))
print('min eigenvalue =', np.linalg.eigvalsh(cov).min())
print('condition number =', np.linalg.cond(cov))
```
